# Diabetes Prediction Using Machine Learning

```python
#Abirami
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score,KFold
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix
import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.get_dummies(df, columns=['gender', 'smoking_history'], drop_first=True,dtype=int)
```

```python
x = df.drop('diabetes',axis=1)
y = df['diabetes']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import RobustScaler
```

```python
scaler = RobustScaler()
```

```python
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
lr = LogisticRegression()
knn = KNeighborsClassifier()
dt = DecisionTreeClassifier()
rf = RandomForestClassifier()
```

```python
lr.fit(x_train,y_train)
knn.fit(x_train,y_train)
dt.fit(x_train,y_train)
rf.fit(x_train,y_train)
```

```python
y_pred_lr = lr.predict(x_test)
y_pred_knn = knn.predict(x_test)
y_pred_dt = dt.predict(x_test)
y_pred_rf = rf.predict(x_test)
```

```python
kfold = KFold(n_splits=10, shuffle=True, random_state=42)
print("Logistic Regression Cross-Validation Scores:", cross_val_score(lr, x, y, cv=kfold).mean())
print("KNN Cross-Validation Scores:", cross_val_score(knn, x, y, cv=kfold).mean())
print("Decision Tree Cross-Validation Scores:", cross_val_score(dt, x, y, cv=kfold).mean())
print("Random Forest Cross-Validation Scores:", cross_val_score(rf, x, y, cv=kfold).mean())
```

```python
from sklearn.metrics import confusion_matrix, classification_report,accuracy_score
```

```python
acc = {
    'Logistic Regression': accuracy_score(y_test, y_pred_lr),
    'KNN': accuracy_score(y_test, y_pred_knn),
    'Decision Tree': accuracy_score(y_test, y_pred_dt), 
    'RandomForestClassifier': accuracy_score(y_test, y_pred_rf)
}
```

```python
acc
```

```python
cm = {
    'Logistic Regression': confusion_matrix(y_test, y_pred_lr),
    'KNN': confusion_matrix(y_test, y_pred_knn),
    'Decision Tree': confusion_matrix(y_test, y_pred_dt),
    'RandomForestClassifier': confusion_matrix(y_test, y_pred_rf)
}
```

```python
cm
```

```python
cr = {
    'Logistic Regression': classification_report(y_test, y_pred_lr),
    'KNN': classification_report(y_test, y_pred_knn),
    'Decision Tree': classification_report(y_test, y_pred_dt),
    'RandomForestClassifier': classification_report(y_test, y_pred_rf)
}
```
